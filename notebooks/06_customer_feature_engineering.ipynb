## Lifecycle stage rules

- Inactive: recency_days > 180
- Loyal/High-Activity: orders >= 3 AND recency <= 180
- Repeat: orders == 2 AND recency <= 180
- New: orders == 1 AND recency <= 90
- Active: orders == 1 AND 90 < recency <= 180
