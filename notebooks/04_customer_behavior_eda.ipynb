# 04 — Customer Behavior EDA

Focused exploratory analysis supporting (not replacing) the customer purchasing-behavior feature engineering that follows.

```python
import pandas as pd
tx = pd.read_csv("../outputs/cleaned/olist_customer_transactions.csv", parse_dates=["order_purchase_timestamp"])
order_level = tx.groupby(["customer_unique_id","order_id"], as_index=False).agg(
    order_purchase_timestamp=("order_purchase_timestamp","min"),
    order_value=("payment_value","first"))
orders_per_customer = order_level.groupby("customer_unique_id")["order_id"].nunique()
orders_per_customer.describe()
```

```python
print(f"1-order customers: {(orders_per_customer==1).sum():,} ({(orders_per_customer==1).mean()*100:.2f}%)")
print(f"2+-order customers: {(orders_per_customer>=2).sum():,} ({(orders_per_customer>=2).mean()*100:.2f}%)")
```

## Monthly order volume
Growth is steady from late 2016 through late 2017, peaking around November 2017 (consistent with a Black Friday effect), continuing into 2018.

```python
monthly = order_level.set_index("order_purchase_timestamp").resample("ME")["order_id"].nunique()
monthly.tail(12)
```
