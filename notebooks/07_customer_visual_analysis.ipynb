```python
import pandas as pd, matplotlib.pyplot as plt
cp = pd.read_csv("../outputs/customer_analysis/customer_behavior_engagement.csv")
tx = pd.read_csv("../outputs/cleaned/olist_customer_transactions.csv", parse_dates=["order_purchase_timestamp"])
print("Visualization source data loaded:", cp.shape, tx.shape)
```

## Chart: Repeat vs One-Time Customers (`09_repeat_vs_onetime_customers.png`)

**Observation:** 97.0% of customers made only one purchase; 3.0% (2,801) are repeat customers.

**Business Interpretation:** The overwhelming majority of the customer base does not return for a second purchase within the observed period.

**Potential Action:** Prioritize post-first-purchase re-engagement rather than assuming organic repeat behavior.

## Chart: Customer Lifecycle Distribution (`14_customer_lifecycle_distribution.png`)

**Observation:** 59.18% of customers (55,252) are Inactive (>180 days since last purchase); only 0.13% reach Loyal/High-Activity status.

**Business Interpretation:** Most of the base has gone cold under the project-defined 180-day threshold.

**Potential Action:** Build a win-back campaign targeting the Inactive segment, prioritized by historical value_segment.

## Chart: Purchase Interval Distribution (`04_purchase_interval_distribution.png`)

**Observation:** Among 2,801 repeat customers, the median interval between purchases is 32 days.

**Business Interpretation:** When customers do return, they tend to do so within roughly a month.

**Potential Action:** Time retention emails/offers around the 25–40 day mark post-purchase.
